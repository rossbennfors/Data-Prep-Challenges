```python
import pandas as pd
import numpy as np
import string
```

```python
df = pd.read_csv('C:\\Users\\RossBennfors\\Documents\\My Tableau Prep Repository\\Datasources\\Prep Challenge Data sets\\PD 2021 Wk 2 Input.csv')
```

```python
df['Model'] = df['Model'].str.replace('\d+', '', regex=True)
```

```python
df['Model'] = df['Model'].str.replace('/', '', regex=True)
```

```python
df['Order Value'] = df['Quantity'] * df['Value per Bike']
```

```python
df_agg1 = df.groupby(['Model', 'Bike Type']).agg(
    Quantity_Sold=('Quantity', 'sum'),
    Avg_Value_per_Bike=('Value per Bike', 'mean'),
    Order_Value=('Order Value', 'sum')
)
```

```python
df_agg1['Avg_Value_per_Bike'] = df_agg1['Avg_Value_per_Bike'].round(1)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df['Shipping Date'] = pd.to_datetime(df['Shipping Date'], format='%d/%m/%Y')
```

```python
df.info()
```

```python
df['Days to Ship'] = (df['Shipping Date'] - df['Order Date']).dt.days
```

```python
df_agg2 = df.groupby(['Store', 'Model']).agg(
    Quantity_Sold=('Quantity', 'sum'),
    Avg_Days_to_Ship=('Days to Ship', 'mean'),
    Order_Value=('Order Value', 'sum')
)
```

```python
df_agg2['Avg_Days_to_Ship'] = df_agg2['Avg_Days_to_Ship'].round(1)
```
