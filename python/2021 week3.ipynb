```python
import pandas as pd
import numpy as np
```

```python
df_combined = pd.concat(
    pd.read_excel(r"C:\Users\RossBennfors\Documents\My Tableau Prep Repository\Datasources\Prep Challenge Data sets\PD 2021 Wk 3 Input.xlsx", sheet_name=None)
)
```

```python
df_combined = df_combined.reset_index(level=0).rename(columns={'level_0': 'Sheet Name'}).reset_index(drop=True)
```

```python
df_melted = df_combined.melt(id_vars=['Date', 'Sheet Name'], var_name='Type_Product', value_name='Value')
```

```python
df_melted[['Type', 'Product']] = df_melted['Type_Product'].str.split(' - ', expand=True)
```

```python
df_melted.drop(columns=['Type_Product'], inplace=True)
```

```python
df_melted['quarter'] = df_melted['Date'].dt.to_period('Q')
```

```python
df_melted['quarter'] = df_melted['quarter'].dt.quarter
```

```python
df_melted
```

```python
df_OP1 = df_melted.groupby(['Product', 'quarter'])['Value'].sum().reset_index()
```

```python
df_OP2 = df_melted.groupby(['Product', 'Sheet Name', 'Type'])['Value'].sum().reset_index()
```

```python
df_OP2['Store'] = df_OP2['Sheet Name']
df_OP2.drop(columns=['Sheet Name'], inplace=True)
```
